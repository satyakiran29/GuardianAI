```python
import plotly.express as px #pip install plotly
import pandas as pd

df = pd.read_csv("Woman_Safety_Dataset_Management.csv")

fig = px.line(
    df,
    x="city",
    y="crime_count",
    title="cimes per city"
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Convert timestamp to datetime
df["incident_timestamp"] = pd.to_datetime(df["incident_timestamp"])

fig = px.line(
    df.sort_values("incident_timestamp"),
    x="incident_timestamp",
    y=["crime_count", "safety_score"],
    title="Crime Count and Safety Score Over Time",
    markers=True
)

fig.show()
```

```python
import plotly.express as px #pip install plotly
import pandas as pd

df = pd.read_csv("Woman_Safety_Dataset_Management.csv")
city_df = (
    df.groupby("city", as_index=False)["crime_count"]
      .mean()
)
fig = px.line(
    city_df,
    x="city",
    y="crime_count",
    color_discrete_sequence=["red"]
)

fig.show()
```

```python
import plotly.express as px

city_df = (
    df.groupby("city", as_index=False)["crime_count"]
      .mean()
)

fig = px.line(
    city_df,
    x="city",
    y="crime_count",
    markers=True,
    title="Average Crime Count by City"
)

fig.show()
```

```python
import plotly.express as px

city_df = (
    df.groupby("city", as_index=False)["crime_count"]
      .mean()
)

fig = px.scatter(
    city_df,
    x="city",
    y="crime_count",
    
    title="Average Crime Count by City"
)

fig.show()
```
