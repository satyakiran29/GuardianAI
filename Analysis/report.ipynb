```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Woman_Safety_Dataset_Management.csv")

# Filter Hyderabad
hyd = df[df["city"] == "Hyderabad"]

# Total crime count by area
area_crimes = hyd.groupby("area")["crime_count"].sum()

# Plot
plt.figure(figsize=(10, 6))
area_crimes.plot(
    kind="bar",
    color="skyblue",
    edgecolor="black"
)

plt.title("Crime Count by Area in Hyderabad")
plt.xlabel("Area")
plt.ylabel("Crime Count")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Woman_Safety_Dataset_Management.csv")

# Convert timestamp to datetime
df["incident_timestamp"] = pd.to_datetime(df["incident_timestamp"])

# Create Month column
df["Month"] = df["incident_timestamp"].dt.month_name()

# Arrange months in calendar order
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

# Sum crime counts by month
monthly_crimes = (
    df.groupby("Month")["crime_count"]
      .sum()
      .reindex(month_order)
)

# Plot bar chart
plt.figure(figsize=(10, 5))
monthly_crimes.plot(
    kind="bar",
    color="skyblue",
    edgecolor="black"
)

plt.title("Monthly Crime Count")
plt.xlabel("Month")
plt.ylabel("Total Crime Count")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd


df = pd.read_csv("Woman_Safety_Dataset_Management.csv")

top_risk_localities = (
    df.groupby(["city", "area"])
    .agg(
        total_logs=("incident_id", "count"),
        avg_safety_score=("safety_score", "mean"),
        high_critical_logs=(
            "risk_level",
            lambda x: x.isin(["High", "Critical"]).sum(),
        ),
        avg_police_dist_km=("police_station_distance_km", "mean"),
        avg_lighting=("lighting_score", "mean"),
    )
    .sort_values(by="avg_safety_score", ascending=True)
    .head(10)
    .reset_index()
)


top_risk_localities.columns = [
    "City",
    "Local Area",
    "Total Logs",
    "Avg Safety Score",
    "High/Critical Logs",
    "Avg Police Dist (km)",
    "Avg Lighting",
]


print(top_risk_localities.to_string(index=False))
```

```python
import pandas as pd


df = pd.read_csv("Woman_Safety_Dataset_Management.csv")

crime_risk_dist = pd.crosstab(df["crime_type"], df["risk_level"])[
    ["Critical", "High", "Medium", "Low"]
]

crime_risk_dist["Total Count"] = crime_risk_dist.sum(axis=1)


crime_risk_dist = crime_risk_dist.sort_values(
    by="Total Count", ascending=False
).reset_index()
crime_risk_dist.rename(columns={"crime_type": "Crime Category"}, inplace=True)


print(crime_risk_dist.to_string(index=False))
```

```python
import pandas as pd


df = pd.read_csv("Woman_Safety_Dataset_Management.csv")


time_risk_dist = pd.crosstab(df["time_of_day"], df["risk_level"])[
    ["Critical", "High", "Medium", "Low"]
]


time_order = ["Morning", "Afternoon", "Evening", "Night", "Late Night"]
time_risk_dist = time_risk_dist.reindex(time_order)


time_risk_dist["Total Count"] = time_risk_dist.sum(axis=1)


time_risk_dist = time_risk_dist.reset_index()
time_risk_dist.rename(columns={"time_of_day": "Time Window"}, inplace=True)

# Display formatted table
print(time_risk_dist.to_string(index=False))
```

```python
import pandas as pd


df = pd.read_csv("Woman_Safety_Dataset_Management.csv")


city_matrix = (
    df.groupby("city")
    .agg(
        incident_count=("incident_id", "count"),
        total_crime_volume=("crime_count", "sum"),
        avg_crime_per_incident=("crime_count", "mean"),
        avg_safety_score=("safety_score", "mean"),
        avg_lighting_score=("lighting_score", "mean"),
        avg_police_dist_km=("police_station_distance_km", "mean"),
        high_critical_incidents=(
            "risk_level",
            lambda x: x.isin(["High", "Critical"]).sum(),
        ),
    )
    .sort_values(by="incident_count", ascending=False)
    .reset_index()
)


city_matrix.columns = [
    "City",
    "Incident Count",
    "Total Crime Volume",
    "Avg Crime / Incident",
    "Avg Safety Score",
    "Avg Lighting Score",
    "Avg Police Dist (km)",
    "High + Critical Incidents",
]

print(city_matrix.to_string(index=False))
```
